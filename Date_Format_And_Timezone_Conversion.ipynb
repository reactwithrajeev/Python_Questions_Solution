```python
date_formats =[
    '%Y-%m-%d %H:%M:%S',           # 2024-01-15 14:30:00
    '%d/%m/%Y %H:%M:%S',           # 15/01/2024 09:45:30
    '%b %d, %Y %I:%M %p',          # Jan 15, 2024 11:20 AM
    '%Y-%m-%dT%H:%M:%SZ',          # 2024-01-15T16:30:00Z
    '%d-%m-%Y %H:%M',              # 15-01-2024 18:45
    '%Y/%m/%d %H:%M:%S',           # 2024/01/16 08:30:00
    '%d-%b-%Y %H:%M:%S',           # 16-Jan-2024 14:15:30
    '%m/%d/%Y %H:%M:%S',           # 01/16/2024 10:00:00
    '%d.%m.%Y %H:%M',              # 17.01.2024 15:45
    '%B %d, %Y %H:%M',             # January 17, 2024 09:00
    '%d/%m/%Y',                    # 18/01/2024
    '%d %b %Y %I:%M %p',           # 19 Jan 2024 11:30 AM
    '%d-%m-%Y',                    # 20-01-2024
    '%b %d, %Y',                   # Jan 22, 2024
    '%d.%m.%Y %H:%M:%S',           # 22.01.2024 13:45:30
    '%d/%m/%Y %H:%M',              # 23/01/2024 08:15
    '%d-%b-%Y %H:%M',              # 24-Jan-2024 10:30
    '%B %d, %Y %I:%M %p',          # January 25, 2024 12:00 PM
    '%Y-%m-%d',                    # 2024-01-26
    '%d.%m.%Y'                     # 27.01.2024
]
```

```python
def date_parsing(date_string,timezone_str ="IST"):
    if pd.isna(date_string) or date_string=="":
        return pd.NaT
    if isinstance(date_string,datetime):
        return date_string
    date_strng = str(date_string).strip()

    #YYYY-MM-DDTHH:MM:SSZ
    #YYYY-MM-DDTHH:MM:SS+05:30

    if "T" in date_strng and ("+" in date_strng or date_strng.endswith("Z")):
        try:
            dt = pd.to_datetime(date_strng,utc=True)
            ist = pytz.timezone('Asia/Kolkata')
            return dt.astimmezone(ist)
        except:
            pass

    for fmt in date_formats:
        try:
            dt = datetime.strptime(date_strng,fmt)
            tz_map = {
                'IST': 'Asia/Kolkata',
                'UTC': 'UTC',
                'PST': 'US/Pacific',
                'EST': 'US/Eastern',
                'SGT': 'Asia/Singapore'
            }
            source_tz = pytz.timezone(tz_map.get(timezone_str,'UTC'))
            date_with_tz = source_tz.localize(dt)
            ist = pytz.timezone('Asia/Kolkata')
            dt_ist = date_with_tz.astimezone(ist)
            return dt_ist
        except(ValueError,AttributeError):
            continue
            

    
```
