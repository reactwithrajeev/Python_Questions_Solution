# HOW TO IMPORT CSV DATA TO MYSQL USING PYTHON

```python
# pandas 
# sqlalchemy 
# mysql-connector-python

# pip pandas , sqlalchecmy , mysql-connector-python

import pandas as pd 
from sqlalchemy import create_engine
```

```python
df = pd.read_csv(r"C:\Users\Rajeev\Downloads\Ecommerce Data.csv")
```

```python
print(df.info())
```

```python
username = "root"
password = *****
host = "localhost"
port = 3306
db_name = 'zomato'
```

```python
#"mysql+mysqlconnector://username:password@localhost:3306/database_name"


engine = create_engine(f"mysql+mysqlconnector://{username}:{password}@{host}:{port}/{db_name}")
```

```python
df.to_sql(
    name = 'ecom_data', 
    con = engine, 
    if_exists='replace', 
    index=False
)
```

```python
pd.read_sql("SELECT * FROM ECOM_DATA", engine)
```
